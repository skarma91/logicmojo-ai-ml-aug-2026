### np.random.randn(shape)

This generates random number from standard normal distribution. (mean = 0, variance = 1)

A NumPy array of given `shape` is generated with values drawn from standard normal distribution.

```python
np.random.randn(100)
```

```python
np.random.randn(5, 3)
```

### Helper function to plot the histogram

```python
import matplotlib.pyplot as plt

def plot_hist(x, edgecolor='k', nbins=20):
    plt.figure(figsize=(10,8))
    plt.hist(x, edgecolor=edgecolor, bins=nbins)
    plt.xlabel('Values')
    plt.ylabel('Frequency')
    plt.show()
```

```python
### uniformly distributed random values

values = np.random.rand(100000) # random numbers are uniformly distributed b/w 0 and 1

plot_hist(values, 'k', 20)
```

```python
values = np.random.uniform(2, 6, 100000) # random numbers are uniformly distributed b/w 2 and 6

plot_hist(values)
```

```python
### Normally distributed random variable

values = np.random.randn(10000) # the random numbers are normally distributed with mean 0 and std dev 1

plot_hist(values)
```

```python
### mean = 5, std dev = 2

values = np.random.normal(loc=5, scale=2, size=10000) # the random numbers are normally distributed with mean 5 and std dev 2

plot_hist(values)
```

```python
def plot_bar(x):

    from collections import Counter

    count_dict = Counter(x)

    sorted_counts = sorted(count_dict.items())

    values = [v[0] for v in sorted_counts]
    counts = [v[1] for v in sorted_counts]

    plt.figure(figsize=(10,8))
    plt.bar(x=values, height=counts)
    plt.show()
```

```python
values = np.random.randint(1, 50, 10000)  # 10000 random integers between 1 to 49

plot_bar(values)
```

### Simulating random experiments

#### Simulate coin toss

```python
np.random.rand(1)[0]  # random number between 0 and 1
```

```python
### Unbiased coin (equal probabilities of getting head and tail)

### Biased coin (head with probability p and tail with probaility (1-p))

def simulate_cointoss(p = 0.5):
    """ Simulate a coin toss with head probability p """

    if p > 1.0 or p < 0:
        raise ValueError("The probability p should be between 0 to 1")
    
    v = np.random.rand(1)[0]

    if v <= p:
        return 'H'
    else:
        return 'T'    
```

```python
def multiple_cointoss(n, p=0.5):
    """Simulate n coin tosses with head probability p"""

    results = [simulate_cointoss(p) for i in range(n)]

    return results
```

```python
results = multiple_cointoss(10000, p=0.5)
```

```python
Counter(results)
```

```python
plot_bar(results)
```

```python
results = multiple_cointoss(10000, p=0.2)
print(Counter(results))
plot_bar(results)
```

```python
results = multiple_cointoss(1000, p=0.7)
print(Counter(results))
plot_bar(results)
```
