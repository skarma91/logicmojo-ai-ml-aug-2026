### List comprehension

List comprehension in Python is a concise, one-line syntax used to create a new list by iterating over an existing iterable (like a list, tuple, or range). It serves as a shorter, more readable alternative to traditional `for` loops and `.append()` blocks.

**Basic Syntax:**

A list comprehension is always written inside square brackets [] and contains three main components:
```python
new_list = [expression for item in iterable if condition]
```

- expression: The operation or value to add to the new list.
- item: The current variable being pulled from the sequence.
- iterable: The collection you are looping through.
- if condition (Optional): A filter that only includes items evaluating to True.

````python
my_list = [3, 5, 10, 11, 2]

# Create another list with squared values of the my_list

result_list = []

for x in my_list:
    result_list.append(x**2)

result_list
````

````python
result_list = [x**2 for x in my_list]   # list comprehension

result_list
````

````python
values = [3, -2, 5, 12, 9, 25, 32, -4]

# create a new list from the values where the (element + 4)**2-10 > 0

new_list = [x for x in values if (x+4)**2-10 > 0]

new_list
````

The general syntax for using an `if-else` condition inside a list comprehension puts the conditional logic before the for loop component.

```python
new_list = [expression_if_true if condition else expression_if_false for item in iterable]
```

- expression_if_true: The value assigned if the condition is met.
- condition: The test applied to each item.
- expression_if_false: The fallback value assigned if the condition fails.
- for item in iterable: The standard loop that iterates through your collection.

````python
number_list = [-3, 8, 9, -4, 0, 12, 4, 5]

# create another list from the elements of the number list
# condition: in the newlist square the number if it is less than 0 else leave as it is

new_number_list = [x**2 if x < 0 else x for x in number_list]

new_number_list
````

**Key Difference: Trailing if vs. Inline if-else**

- Trailing `if` (Filtering): Placed at the very end to filter elements out of the final list.

- Inline `if-else` (Transforming): Placed before the for loop to choose between two values for every item.

### Dictionary comprehension

Dictionary comprehension in Python is a concise, single-line syntax used to create new dictionaries from an iterable. It serves as a shorter, faster, and more readable alternative to traditional `for` loops.

**General Syntax:**

A dictionary comprehension is enclosed in curly braces {}. It requires both a key expression and a value expression separated by a colon, followed by a for loop:

```python
{key_expression: value_expression for item in iterable if condition}
```

- key_expression: Determines how the dictionary keys are generated.
- value_expression: Determines how the corresponding values are generated.
- if condition (Optional): Filters which items from the iterable are

````python
# Traditional Loop Way
squares = {}
for n in range(1, 6):
    squares[n] = n ** 2

# Dictionary Comprehension Way
squares = {n: n ** 2 for n in range(1, 6)}
print(squares)
````

````python
# Create a dictionary of squares, but only for even numbers
even_squares = {n: n ** 2 for n in range(10) if n % 2 == 0}
print(even_squares)
````

````python
# Convert prices from USD to EUR (assuming 1 USD = 0.92 EUR)
usd_prices = {'apple': 1.00, 'banana': 0.50, 'orange': 0.75}

eur_prices = {item: round(price * 0.92, 2) for item, price in usd_prices.items()}
print(eur_prices)
````

````python
# combining two lists into a dictionary

keys = ['name', 'role', 'language']
values = ['Alice', 'Developer', 'Python']

user_profile = {k: v for k, v in zip(keys, values)}
print(user_profile)
````

Dictionary comprehension with an if-else condition allows you to construct a new dictionary by dynamically altering its keys or values based on a specific condition.Just like list comprehensions, the if-else block must be placed before the for loop loop component.

**General Syntax**

```python
new_dict = {key: (value_if_true if condition else value_if_false) for key, value in iterable}
```

````python
scores = {"Alice": 85, "Bob": 52, "Charlie": 90, "David": 45}

# Syntax in action
results = {name: ("Pass" if score >= 60 else "Fail") for name, score in scores.items()}
````

**Key Difference: Trailing if vs. Inline if-else**

- Trailing `if` (Filtering): Placed at the very end to filter elements out of the final dictionary.

- Inline `if-else` (Transforming): Placed before the for loop to choose between two values for every item.

**When to avoid dictionary and list comprehension**

You should avoid dictionary and list comprehensions whenever they make your code harder to read, maintain, or debug. While they are powerful tools, overusing them can lead to "clever" code that is difficult for humans to understand.

Here are the specific situations where you should skip comprehensions and use traditional loops instead:

1. Complex nested loop
2. Too Many Conditional Statements
3. Large Datasets (Memory Exhaustion)
4. Difficult Debugging
5. Side Effects Inside the Loop
    Comprehensions are meant for creating collections, not for executing actions. If you are printing data, modifying external variables, or writing to a file, do not use them. So **do not use** `[print(x) for x in my_list]` or similar execution inside list or dictionary comprehension.
